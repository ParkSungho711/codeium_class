```python
answer = 7
guess = 0

while guess != answer:
    guess = int(input("1~10 숫자를 맞혀보세요: "))

    if guess < answer:
        print("더 큰 숫자입니다.")
    elif guess > answer:
        print("더 작은 숫자입니다.")

print("정답입니다!")
```

```python
HP = 100

while HP > 0:
    fight = input('때릴까? 말까?? ')

    if '때린다' in fight:
        HP -= 10
        print(f'맞아서 HP가 {HP} 남았습니다.')
    else:
        print('피했습니다.')

print('죽었습니다.')
```

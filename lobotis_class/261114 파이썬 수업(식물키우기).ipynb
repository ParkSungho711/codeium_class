```python
import time

print('=== 🌱 7일 식물 키우기 ===')
print('매일 무엇을 해 줄지 골라주세요~>< \n')

growth = 0
water_streak = 0
for day in range(1,8):
    print(f'--- {day}일차 (성장도 {growth})')
    care = input('무엇을 할까요? (물 / 햇빛 / 그냥둠) : ')

    if care == '물':
        water_streak += 1

        if water_streak >= 3:
            growth -= 20
            water_streak = 0
            print('🦠 물을 3일 연속 주어서, 뿌리가 썩었습니다...')
        else:
            growth += 15
            print('💧 물을 주었습니다. (+15)')
    elif care == '햇빛':
        growth += 10
        print('☀️ 햇빛을 쬐었습니다. (+10)')
    else:
        growth -= 5
        print('😴 아무것도 하지 않았습니다. (-5)')
        
    if growth < 0:
        growth = 0
    # 지금 모습
    if growth >= 70:
        shape = '🌳 큰 나무'
    elif growth >= 40:
        shape = '🌿 줄기'
    elif growth >= 15:
        shape = '🌱 새싹'
    else:
        shape = '🫘 씨앗'


    print(f'{shape} 성장도 {growth}\n')
    time.sleep(0.4)

# 7일 뒤에 결과
print('=== 7일이 지났습니다 ===')
if growth >= 70:
    print('🌳 멋진 나무가 되었습니다! 식물 키우기 성공 !')
elif growth >= 40:
    print('🌿 줄기까지 자랐습니다. 조금 더 돌봐 주세요.')
elif growth >= 15:
    print('🌱 새싹에서 멈췄습니다. 물을 더 자주 줘 보세요.')
else:
    print('🫘 씨앗 그대로 입니다... 다음엔 꼭 물을 주세요...ㅠ')
```

```python
import warnings; warnings.filterwarnings("ignore")
import psycopg2
import pandas as pd
from IPython.display import HTML, display

conn = psycopg2.connect(
    host='localhost',
    port=5432,
    dbname='postgres',
    user='postgres',
    password='changeme'    
)
cur = conn.cursor()
#print('연결완료!')

# 이미 테이블 있으면 지운다
cur.execute("DROP TABLE IF EXISTS scores")
# 테이블 생성
cur.execute("""
   CREATE TABLE scores (
       id      SERIAL PRIMARY KEY,
       name    TEXT,
       subject TEXT,
       score   INT
   )
""")
# 데이터 삽입
data = [
    ("홍길동","수학",92),
    ("홍길동","영어",78),
    ("노원태","수학",88),
    ("노원태","영어",90),
    ("이인성","수학",70),
    ("이인성","영어",80),
    ("김이안","수학",100),
    ("김이안","영어",100),
]

cur.executemany(
    'INSERT INTO scores (name, subject, score) values (%s, %s, %s)',
    data
)

conn.commit()

print(f'{len(data)}줄 저장 완료')
```

```python
df = pd.read_sql("SELECT name, subject FROM scores WHERE score >= 90 ORDER BY score DESC", conn)
display(HTML(df.to_html(index=False)))
```

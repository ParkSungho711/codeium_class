```python
import time
import random

print('=== 🎣 낚시 게임 (5번 던지기) ===')

# 횟수
tries = 5
# 점수
score = 0
# 잡은 물고기 담을 가방
bag = []

while tries > 0:
    input(f'\n남은 횟수 {tries}번, 엔터를 누르면 낚싯대를 던집니다.')

    print('🎣휙~~~! 기다리는 중2정헌', end='')

    for i in range(3):
        time.sleep(0.5)
        print('.', end='',flush=True)
    print()
# 확률 만들것이다
    n = random.randint(1,10)
# 40% 멸치  30% 상 10% 낡은장화 10% 복어
# 10 황금물고기
    if n <= 4:
        fish = '🐟 멸치'
        point = 1
    elif n <= 7:
        fish = '🦈 상어'
        point = 3
    elif n == 8:
        fish = '👢 낡은 장화'
        point = -1
    elif n == 9:
        fish = '🪼 해파리'
        point = -2
    else:
        fish = '👑 황금물고기'
        point = 10

    bag.append(fish)
    score = score + point
    tries -= 1

    print(f'{fish} 낚시 성공! ({point})점, 현재 점수 {score}')

# 반복은 끝난거야
print('\n=== 낚시 끝~!! ===')
print(f'잡은 것 {len(bag)}개 : {bag}')

kinds = ['🐟 멸치', '🦈 상어','👢 낡은 장화', '🪼 해파리', '👑 황금물고기']
for k in kinds:
    if bag.count(k) > 0:
        print(f'    {k} x {bag.count(k)}')
# 🏆🥇🥈🥉👢
print(f'\n최종 점수 : {score}점')
if score >= 15:
    print('🏆 낚시왕 !!!!')
elif score >= 8:
    print('🥇 프로 낚시꾼!')
elif score >= 3:
    print('🥈 그냥 낚시꾼!')
else:
    print('👢 장화 수집가!')
```
